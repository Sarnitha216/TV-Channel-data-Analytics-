#DATA CLEANING 

```python
#DATA Cleaning Standardize channel names
df["Channel_Name"] = (
    df["Channel_Name"]
    .str.strip()
    .str.title()
    .str.replace("Tv", "TV", regex=False)
)

print(df["Channel_Name"])
```

```python
# Handling missing values

# Numerical columns → median
for col in df.select_dtypes(include="number").columns:
    df[col] = df[col].fillna(df[col].median())

# Categorical columns → mode
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].fillna(df[col].mode()[0])

print(df.isnull().sum())
```

```python
# Remove duplicate records

print("Duplicates before:", df.duplicated().sum())

df = df.drop_duplicates()

print("Duplicates after:", df.duplicated().sum())
```

```python
# Check invalid Avg_Rating values

invalid_ratings = df[
    (df["Avg_Rating"] < 0) | (df["Avg_Rating"] > 5)
]

print("Number of invalid ratings:", len(invalid_ratings))
print(invalid_ratings)
```

```python
# Correct invalid Avg_Rating values

valid_median = df.loc[
    (df["Avg_Rating"] >= 0) & (df["Avg_Rating"] <= 5),
    "Avg_Rating"
].median()

df.loc[
    (df["Avg_Rating"] < 0) | (df["Avg_Rating"] > 5),
    "Avg_Rating"
] = valid_median

print("Invalid ratings corrected.")
```

```python
# Verify invalid ratings

invalid_ratings = df[
    (df["Avg_Rating"] < 0) | (df["Avg_Rating"] > 5)
]
print("Number of invalid ratings:", len(invalid_ratings))
```

```python
print("Minimum Rating:", df["Avg_Rating"].min())
print("Maximum Rating:", df["Avg_Rating"].max())
```

```python
# Correct negative Subscription_Fee values
df.loc[df["Subscription_Fee"] < 0, "Subscription_Fee"] = df["Subscription_Fee"].median()
print(df["Subscription_Fee"].min())
```

```python
# Fill ratings with median
df["Avg_Rating"] = df["Avg_Rating"].fillna(df["Avg_Rating"].median())

print("Avg Rating")
print(df["Avg_Rating"])

df["Subscription_Fee"] = df["Subscription_Fee"].fillna(
    df["Subscription_Fee"].median()
)

print("Subscription Fee")
print(df["Subscription_Fee"])

print("\nMinimum Subscription Fee")
print(df["Subscription_Fee"].min())
```

```python
# Handling unrealistic Years_Active values
df["Years_Active"] = df["Years_Active"].fillna(
    df["Years_Active"].median()
)

print(df["Years_Active"].min())
```

```python
# Fill numerical missing values
for col in df.select_dtypes(include="number").columns:
    df[col] = df[col].fillna(df[col].median())
```

```python
# Convert numeric columns stored as text into numeric format
num_cols = ["Avg_Viewers", "Total_Likes", "Social_Media_Followers", "YouTube_Views", "Monthly_Reach"]
for col in num_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")
print(df[num_cols].dtypes)
```

```python
# Replace invalid text values with missing values
df = df.replace(
    ["NA", "N/A", "na", "n/a", "unknown", "Unknown", ""],pd.NA)
print(df.isnull().sum())
```

```python
# Fill numerical missing values with median
for col in df.select_dtypes(include="number").columns:
    df[col] = df[col].fillna(df[col].median())

# Fill categorical missing values with mode
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].fillna(df[col].mode()[0])

print(df.isnull().sum())
```

```python
print("Missing values:")
print(df.isnull().sum())
print("\nDuplicate records:")
print(df.duplicated().sum())
```

#CHANNEL-WISE ANALYSIS

```python
# Number of records for each channel
print("===== NUMBER OF RECORDS =====")
print(df["Channel_Name"].value_counts())

# Average viewers by channel
print("\n===== AVERAGE VIEWERS =====")
print(df.groupby("Channel_Name")["Avg_Viewers"].mean())

# Average rating by channel
print("\n===== AVERAGE RATING =====")
print(df.groupby("Channel_Name")["Avg_Rating"].mean())

# Average likes by channel
print("\n===== AVERAGE LIKES =====")
print(df.groupby("Channel_Name")["Total_Likes"].mean())

# Top 5 channels by average viewers
print("\n===== TOP 5 CHANNELS BY VIEWERS =====")
print(df.groupby("Channel_Name")["Avg_Viewers"].mean().sort_values(ascending=False).head(5))
```

#SHOW CATEGORY ANALYSIS

```python
# Number of records in each category
print("===== NUMBER OF RECORDS =====")
print(df["Show_Category"].value_counts())

# Average viewers by category
print("\n===== AVERAGE VIEWERS =====")
print(df.groupby("Show_Category")["Avg_Viewers"].mean())

# Average rating by category
print("\n===== AVERAGE RATING =====")
print(df.groupby("Show_Category")["Avg_Rating"].mean())

# Average likes by category
print("\n===== AVERAGE LIKES =====")
print(df.groupby("Show_Category")["Total_Likes"].mean())

# Highest average viewers category
print("\n===== HIGHEST VIEWED CATEGORY =====")
print(df.groupby("Show_Category")["Avg_Viewers"].mean().idxmax())
```
